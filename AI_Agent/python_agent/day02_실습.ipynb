```python
# 삼항 표현식 - if/else를 한 줄로
age = 20
status = "성인" if age >= 18 else "미성년"
print(status)      # 성인

# 범위 비교를 한 번에 (조건 체이닝)
score = 85
if 80 <= score < 90:
    print("B등급")   # 80 이상 90 미만

# 여러 조건을 논리 연산자로 결합
is_weekend, is_holiday = True, False
if is_weekend or is_holiday:
    print("쉬는 날")
```

```python
score = 72

# [1번 문제]
# if, elif, elif, else 를 이용해 점수에 맞는 등급을 판별하세요.
# 90 이상 A, 80 이상 B, 70 이상 C, 그 미만은 D 입니다.
# 판별한 등급을 grade 변수에 저장하세요.

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
else:
    grade = 'D'

print(grade)

# [2번 문제]
# f-string으로 '점수 72점, 등급은 C입니다' 형태로 출력하세요.

print(f"점수 {score} 점, 등급은 {grade}입니다")

# [도전 문제]
# 삼항 표현식을 이용해 60점 이상이면 '합격', 미만이면 '불합격'이
# 되는 result 변수를 한 줄로 만들어보세요.

result = "합격" if score >= 60 else "불합격"
print (result)  
```

```python
nums = [1, 2, 3, 4, 5]

# 기존 방식 (3줄)
squares = []
for n in nums:
    squares.append(n * n)

# 리스트 컴프리헨션 (한 줄로 축약)
squares2 = [n * n for n in nums]
print(squares2)              # [1, 4, 9, 16, 25]

evens = [n for n in nums if n % 2 == 0]
print(evens)                 # [2, 4]
```

```python
dan = 7

# [1번 문제]
# for문과 range(1, 10)을 이용해
# '7 x 1 = 7' 형태로 9줄을 출력하세요.

for i in range(1,10):
    print(f"{dan} x {i} =",dan*i) 

print()
# [2번 문제]
# while문으로도 똑같은 결과를 만들어보세요.

i = 1
while(i<10):
    print(f"{dan} x {i} =",dan*i)
    i=i+1 

# [도전 문제]
# 리스트 컴프리헨션으로 [7, 14, 21, ..., 63] 처럼
# 구구단 결과만 담긴 리스트를 한 줄로 만들어보세요.
arr = [dan * i for i in range(1,10)]
print(arr)
```

```python
# [1번 문제]
# 몸무게(kg)와 키(m)를 매개변수로 받아 BMI(체질량지수)를 반환하는
# 함수 calc_bmi(weight, height)를 만드세요.
# BMI = 몸무게 / (키 * 키)

def calc_bmi(weight, height):
   return weight / (height * height)

# [2번 문제]
# calc_bmi(70, 1.75) 를 호출해 결과를 출력하세요.

print(calc_bmi(70, 1.75))

# [도전 문제]
# 삼항 표현식으로 18.5 미만 '저체중', 25 이상 '과체중',
# 그 사이는 '정상'을 함께 출력해보세요.
bmi = calc_bmi(70, 1.75)
status = "저체중" if bmi < 18.5 else ("과체중" if bmi >= 25 else "정상")
print(f"BMI: {bmi:.2f}, 결과: {status}")

```

```python
import random

# [1번 문제]
# random.randint(1, 45) 로 로또 번호 1개를 뽑아 출력하세요.
print(random.randint(1, 45))
# [2번 문제]
# for문을 이용해 번호 6개를 뽑아 리스트에 담아 출력하세요.
# (힌트: 빈 리스트를 만들고 append로 추가)
arr=[]

for _ in range(6):
    arr.append(random.randint(1,45))

print(arr)

# [도전 문제]
# 리스트 컴프리헨션으로 2번 문제를 한 줄로 다시 작성해보세요.
# 예: [random.randint(1, 45) for _ in range(6)]

arr=[random.randint(1, 45) for _ in range(6)]

print(arr)
```

```python
# 파일에 쓰기: “w” 쓰기모드로 파일을 열어줌 -> 파일이 없으면 새로 생성하고, 있으면 덮어 쓴다.
with open("memo.txt", "w") as f: # 파일을 f 변수에 저장해서 사용
    f.write("오늘 수업 완료")

# 파일 읽기: “r” 읽기모드로 파일을 열어줌 -> 파일이 없으면 에러 발생
with open("memo.txt", "r") as f:
    content = f.read()
    print(content)        # 오늘 수업 완료
```
