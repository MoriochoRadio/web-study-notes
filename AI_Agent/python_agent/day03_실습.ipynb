```python
data = {
    "name": ["철수", "영희", "영희", "민수"],
    "score": [88, 95, 95, None],
}
df = pd.DataFrame(data)

# [1번 문제] df.isna().sum() 으로 결측치 개수를 확인하세요.
print(df.isna().sum())
print()

# [2번 문제] fillna(0) 으로 결측치를 채운 결과를 출력하세요.
print(df.fillna(0))       # print() 안에 바로 넣어 출력!
print()

# [도전 문제] drop_duplicates() 로 중복된 행을 제거해보세요.
print(df.drop_duplicates()) # print() 안에 바로 넣어 출력!
```

```python
print(df)

df.rename(columns={"name": "이름"})   # 컬럼명 변경
df.drop(columns=["score"])            # 컬럼 삭제


# apply()로 조건에 따라 새 컬럼 만들기
# "grade" 컬럼 생성, "score"컬럼에 값을 x로 하나씩 받아서 조건에 맞는 "A" 나 "B"값을 생성해서 추가
df["grade"] = df["score"].apply(
    lambda x: "A" if x >= 90 else "B"
)

print(df)
```

```python
data = {"name": ["철수", "영희"], "score": [88, 95]}
df = pd.DataFrame(data)
print(df ,"\n")
# [1번 문제] rename() 으로 'name' 컬럼을 '이름'으로 바꾸세요.
df = df.rename(columns={"name": "이름"})

# [2번 문제] apply()와 삼항 표현식을 이용해 score가 90 이상이면
# 'A', 아니면 'B'인 grade 컬럼을 새로 추가하세요.
df["grade"] = df["score"].apply(
    lambda x: "A" if x >= 90 else "B"
)
print(df ,"\n")
# [도전 문제] drop() 으로 score 컬럼을 삭제한 결과를 출력하세요.
df = df.drop(columns=["score"]) 
print(df ,"\n")
```

```python
data = {
    "name": ["철수", "영희", "영희", "민수", "지현"],
    "score": [88, None, 95, 60, 72],
}

# [1번 문제] DataFrame으로 만들고 fillna(0)으로 결측치를 채우세요.
df = pd.DataFrame(data)
df = df.fillna(0)
print(df, "\n")

# [2번 문제] drop_duplicates() 로 중복 행을 제거하세요.
df = df.drop_duplicates()
print(df, "\n")
# 점수가 달라서 영희 중복으로 인식 안됨, 그래도 제거하고싶으면 subset이용해서 처리
df=df.drop_duplicates(subset=['name'],keep='last')
print(df)


# [3번 문제] score가 70 이상인 행만 필터링해 출력하세요.
print(df[df["score"] >= 70],"\n")


# [도전 문제] apply()로 grade 컬럼을 추가하고, 최종 결과를
# to_csv('result.csv', index=False) 로 저장해보세요.
df["grade"] = df["score"].apply(
    lambda x: "A" if x >= 90 else "B"
)

df.to_csv("result.csv", index=False, encoding="utf-8-sig")

print("[최종 DataFrame]")
print(df)
```
