```python
from langchain_core.tools import tool

# [1번 문제]
# 사칙연산 문자열(예: "3*4+2")을 계산해 반환하는 calculate(expression: str) 함수
@tool
def calculate(expression: str) -> str:
    """사칙연산 수식 문자열(예: '3*4+2')을 계산해 결과를 문자열로 반환합니다."""
    try:
        # 문자열로 된 수식을 파이썬이 직접 계산하도록 eval() 사용
        result = eval(expression)
        return str(result)
    except Exception as e:
        return f"계산 오류: {e}"


# [2번 문제]
# 단어를 받아 글자 수를 반환하는 count_letters(word: str) 함수
@tool
def count_letters(word: str) -> int:
    """단어를 입력받아 글자 수(길이)를 반환합니다."""
    return len(word)


# [도전 문제]
# 두 Tool의 .name과 .description을 각각 출력해서 docstring 확인
print("=== 1. calculate Tool 정보 ===")
print("이름 (name)        :", calculate.name)
print("설명 (description) :", calculate.description)

print("\n=== 2. count_letters Tool 정보 ===")
print("이름 (name)        :", count_letters.name)
print("설명 (description) :", count_letters.description)

print(calculate.invoke({'expression':'3*4+2'}))
print(count_letters.invoke({'word':'안녕하세요'}))
```

```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.output_parsers import StrOutputParser
from langchain_core.messages import HumanMessage, AIMessage

# 환경변수 로드 및 LLM 준비
load_dotenv()
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)


# ==============================================================================
# [1번 문제] 
# system 페르소나와 이전 대화를 기억할 MessagesPlaceholder("history")를 포함한 프롬프트 생성
# ==============================================================================
prompt = ChatPromptTemplate.from_messages([
    ("system", "너는 친절한 수학 선생님이야. 학생의 질문에 차근차근 풀이 과정을 설명해줘."),
    MessagesPlaceholder(variable_name="history"),  # 이전 대화 목록이 리스트 형태로 들어오는 자리
    ("human", "{input}")                           # 현재 사용자의 질문
])


# ==============================================================================
# [2번 문제]
# 체인 구성 후 calculate() 결과(14)를 포함해 LLM에게 결과 검증 요청
# ==============================================================================
# 1) 체인 생성 (Prompt -> LLM -> 문자열 파서)
chain = prompt | llm | StrOutputParser()

# 2) 대화 기록을 보관할 history 리스트 초기화
history = []

# 3) 앞서 만든 calculate Tool로 3*4+2 계산 수행
calc_result = calculate.invoke({"expression": "3*4+2"})  # 결과: '14'

# 4) 계산 결과를 질문에 포함하여 질문 생성
question_1 = f"3 * 4 + 2를 계산했더니 {calc_result} 값이 계산 결과인데 이 결과가 맞는지 설명해줘"

# 5) 체인 실행 (대화 기록 history와 질문 input 전달)
response_1 = chain.invoke({
    "history": history,
    "input": question_1
})

print("=== [2번 문제] 수학 선생님 답변 ===")
print(response_1)

# 6) 다음 대화에서 기억할 수 있도록 history 리스트에 이전 질문과 답변 저장
history.append(HumanMessage(content=question_1))
history.append(AIMessage(content=response_1))


# ==============================================================================
# [도전 문제]
# 이전 계산 결과(14)를 기억하는지 확인하기 위해 같은 history로 연계 질문
# ==============================================================================
question_2 = "방금 계산한 값에 10을 더하면?"

response_2 = chain.invoke({
    "history": history,
    "input": question_2
})

print("\n=== [도전 문제] 대화 기억 확인 답변 ===")
print(response_2)
```
