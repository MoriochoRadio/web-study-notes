```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print(arr)          # [1 2 3 4 5]
print(type(arr))    # numpy.ndarray
```

```python
arr = np.array([1, 2, 3, 4, 5])

print(arr * 2)          # [ 2  4  6  8 10]  - 전체에 곱셈
print(arr[arr > 2])     # [3 4 5]           - 조건 필터링

print(arr.mean())       # 3.0   평균
print(arr.max())        # 5     최댓값
print(arr.sum())        # 15    합계
```

```python
scores = np.array([88, 92, 75, 60, 95])

# [1번 문제]
# scores 배열의 평균(mean), 최댓값(max), 최솟값(min)을
# 각각 출력하세요.
print(scores.mean())    
print(scores.max())        
print(scores.min())
# [2번 문제]
# scores 배열에서 80점 이상인 값만 필터링해 출력하세요.
print(scores[scores > 2])
# [도전 문제]
# 모든 값에 5점을 더한 새 배열을 만들어 출력하세요.
newarr=scores+5
print(newarr)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 한글 폰트 설정
plt.rc("font", family="Malgun Gothic")
plt.rc("axes", unicode_minus=False)

scores = [55, 68, 72, 80, 85, 90, 95, 60, 77, 83]

# [1번 문제]
# sns.histplot() 으로 scores의 분포를 히스토그램으로 그리세요.
sns.histplot(scores)
plt.savefig('check_13a.png')
plt.title("성적 히스토그램")
plt.show()

# [2번 문제]
# sns.boxplot() 으로 scores의 박스플롯을 그리세요.
sns.boxplot(y=scores)
plt.title("성적 박스플롯")
plt.show() 

# [도전 문제]
# 임의의 hours(공부 시간) 리스트를 만들고 sns.scatterplot()으로
# hours와 scores의 관계를 산점도로 그려보세요.


hours = [2, 3, 4, 5, 6, 7, 8, 3, 5, 6]

sns.scatterplot(x=hours, y=scores)
plt.title("공부 시간과 성적의 관계")
plt.xlabel("공부 시간(시간)")
plt.ylabel("성적")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# 한글 폰트 및 마이너스 부호 설정 (윈도우)
plt.rc("font", family="Malgun Gothic")
plt.rc("axes", unicode_minus=False)

data = {
    "name": ["철수", "영희", "민수", "지현", "수진"],
    "score": [88, 95, 60, 72, 81],
}

# ==========================================
# [1번 문제] DataFrame으로 만들고 describe() 로 통계를 확인하세요.
# ==========================================
df = pd.DataFrame(data)
print("[1번 문제: describe 통계]")
print(df.describe())
print()

# ==========================================
# [2번 문제] score를 .values 로 Numpy 배열로 바꿔 평균과 표준편차(np.std)를 구하세요.
# ==========================================
score_arr = df["score"].values  # Numpy 배열로 변환

mean_val = np.mean(score_arr)  # 또는 score_arr.mean()
std_val = np.std(score_arr)

print("[2번 문제: Numpy 계산]")
print(f"평균: {mean_val:.2f}")
print(f"표준편차: {std_val:.2f}")
print()

# ==========================================
# [3번 문제] sns.barplot(x="name", y="score", data=df) 로 그리세요.
# ==========================================
sns.barplot(x="name", y="score", data=df)
plt.title("학생별 성적")
plt.xlabel("이름")
plt.ylabel("점수")
plt.show()

# ==========================================
# [도전 문제] score가 80 이상인 사람만 필터링해 새로 그려보세요.
# ==========================================
df_over_80 = df[df["score"] >= 80]

sns.barplot(x="name", y="score", data=df_over_80)
plt.title("80점 이상 학생 성적")
plt.xlabel("이름")
plt.ylabel("점수")
plt.show()
```
